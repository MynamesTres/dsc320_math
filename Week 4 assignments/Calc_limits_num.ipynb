# Calculating Limits Numerically 

let f(x) = 2x^3 - 4x + 1 and let g(x) = e^x - 1 / x

(a) Use Python code to evaluate the function f at appropriate values to calculate lim f(x). x→3 

- Hint: Evaluate 'f' at values closer and closer to 3 and see what value the outputs get close to.

```python
# Writing a function to translate the math formula into Python 
def f(x):
    return 2*x**3 - 4*x + 1 

# Approaching values for x = 3 from both sides 
values = [
    2.9, 2.99, 2.999, 2.9999,
    3.0001, 3.001, 3.01, 3.1
]

# Printing f(x) at each value to see what it converges to
for x in values: 
    print(f"f({x}) = {f(x):.3f}")
```

(b) Use Python code to evaluate the function g at appropriate values to calculate lim g(x). x -> 0

```python
# g(x)
def g(x):
    return (np.exp(x) - 1) / x

# Approaching values for x = 0 from both sides 
values = [
    -0.1, -0.01, -0.001, -0.0001, 
    0.0001, 0.001, 0.01, 0.1
]

# Printing f(x) at each value to see what it converges to
for x in values: 
    print(f"g({x}) = {g(x):.4f}")
```

(c) Could the limits in parts (a) and (b) be calculated by plugging in the value where the limit is taken
into the function?        

Explain. What does this mean in terms of continuity?

**Answer:** 
- For f(x): yes, when plugging in x = 3 it gives us 43 directly. This means that if f is continuous at x = 3, the limit equals the function value. 

- As for g(x), g(0) is undefined because plugging in x = 0 directly causes division by zero. This means that if g is discontinuous at x = 0. The limit does exist and equals 1, but still the function cannot be evaluated at that exact point, which is why we use the numerical approach.

```python
# Testing what happens if we plug in directly
print(f"f(3) = {f(3):.3f}")

# for g(x) at x = 0 
print(g(0)) 
```

# Average Rate of Change      
The formula for the average rate of change of a function f between two values a and b is given by:    

Average Rate of Change = (f(b) - f(a)) / (b - a)    

Create a Python function that takes in a mathematical function, e.g., f(x) = 3x^2 and two numbers a and b
and returns the average rate of change of the function between a and b.


```python
# Writing a function to translate the math formula into Python 
def f(x):
    return 3*x**2

# The average rate of change function 
def average_rate_of_change(func, a, b):
    return (func(b) - func(a)) / (b - a)

# Testing with example function f(x) = 3x^2 to verify the formula works
print("The average rate of change between these two are:", average_rate_of_change(f, 2, 5))
```

# Average Rate of Change to Instantaneous Rate of Change 

A baseball is dropped from a tall cliﬀ. Neglecting air resistance, the distance traveled by the baseball in meters after t seconds is given by the function

f(t) = 4.9t^2

```python
# Writing a function to translate the math formula into Python 
def f(t):
    return 4.9*t**2 

# Average Rate of Change function 
def average_rate_of_change(func, a, b):
    return (func(b) - func(a)) / (b - a)
```

(a) Find the average speed (average rate of change of distance) of the baseball between 5 and 6 seconds.


```python
# (a) Average speed between 5 and 6 seconds - interval of 1 second
print("(a)", average_rate_of_change(f, 5, 6))
```

(b) Find the average speed (average rate of change of distance) of the baseball between 5 and 5.5 seconds.


```python
# (b) Average speed between 5 and 5.5 seconds - interval shrinks to 0.5 seconds
print("(b)", average_rate_of_change(f, 5, 5.5))
```

(c) Find the average speed (average rate of change of distance) of the baseball between 5 and 5.1 seconds.

```python
# (c) Average speed between 5 and 5.1 seconds - interval shrinks to 0.1 seconds
print("(c)", average_rate_of_change(f, 5, 5.1))
```

(d) What is the instantaneous speed of the baseball at t = 5 seconds? Hint: You might want to do a few more calculations similar to those done in parts (a) - (c).

```python
# (d) Instantaneous speed of a baseball between 5, 5.01, and 5, 5.001
# Shrinking the interval towards 0 - the values converge to 49.0, which is the instantaneous speed at t = 5 seconds
print("(d1)", average_rate_of_change(f, 5, 5.01))
print("(d2)", average_rate_of_change(f, 5, 5.001))
```

(e) Find the derivative of f. Hint: The derivative of a quadratic function f(x) = ax^2  + bx + c is given by f(x) = 2ax + b.

```python
# (e) Since a = 4.9 and b = 0 , the derivative simplifies to f'(t) = 2 * 4.9 * t 
def f_derivative(t):
    return 2 * 4.9 * t 
```

(f) Evaluate the derivative of f at t = 5. How does this value compare to the value found in part (d)?     

Explain what is happening.

**Answer**      
This matches the value we found numerically in part (d): both methods arrive at 49.0, confirming that the derivative is the instantaneous rate of change. 


```python
# (f) Evaluate the derivative at t = 5, with a result of 49.0, which matches part (d) 
# This confirms that the derivative is the instantaneous rate of change. 
print("(f)", f_derivative(5))
```

# Calculating and Interpreting Partial Derivatives

A multivariate linear model is trained to predict the selling price of a particular used car model. The linear model predicts the selling price (P ) in US dollars, using its current condition (C) on a scale of 1 − 10 and the age of the car in years (Y ). The model is given by:     

P = 16,000 + 2,400C - 1800Y

(a) What is this model’s predicted selling price of a 5-year old car with a condition rating 8?

```python
# P = price | C = condition | Y = year 

# Writing a function to translate the math formula into Python 
def P(C, Y):
    return 16000 + 2400*C - 1800*Y

# Predict selling price 
print(f"(a) ${P(8, 5)}.00" ) 
```

(b) Find ∂P/∂C and interpret the value 

```python
# (b) Finding partial (P) / partial (C)

 # Writing a function to translate the math formula into Python 
def partial_c(): 
    return 2400

print(f"(b) - For every 1 point increase in rating, the car value increases by ${partial_c()}.00")
```

(c) Find ∂P/∂Y and Interpret the vlaue 

```python
# (c) Finding partial (P) / partial  (Y)

 # Writing a function to translate the math formula into Python 
def partial_y(): 
    return -1800

print(f"(c) - For every additional year the value of the car decreases by ${abs(partial_y())}.00")
```
