```python
# Importing libaries  
import numpy as np 
import matplotlib.pyplot as plt 
```

**Python Function** 

Write a Python function to take in a mathematical function
f, two interval endpoints a and b, and the number of rectangles n, and then outputs the right Riemann sum Rn.

```python
# Right Riemann Sum 
def right_sum(f, a, b, n):
    # Δx -> Delta x 
    dx = (b - a)/n
    # Riemann sum function 
    total = sum(f(a + k * dx) for k in range (1, n + 1))
    
    # Return total 
    return total * dx 

```

**Sketch a graph of f** 

(a) Sketch a graph of f and the rectangles that make up the right Riemann sum R4 on the interval [0, 2].
This can be done by hand.




**Solving for R4**

(b) Use the function you created in the previous problem to calculate R4.

```python
# Area under the curve function 
f = lambda x: np.sqrt(4 - x**2)

# Setting up right Riemann sum 
a = 0 
b = 2
n = 4
dx = (b - a) / n 

result = right_sum(f, a, b, n)
print(f"R4 = {result:.4f}")
```

**Calculate appropriate right Riemann Sums**




**Answer:**

 As (n) increases, (Rn) approaches approximately 3.14. The special name for this value is "pi". 

```python
for n in [10, 100, 1000, 10000, 100000]:
    print(f"{n} = {right_sum(f, 0, 2, n):.4f}")
```

```python
# X_values from 0 -> 2 for the smooth curve
x_values = np.linspace(0, 2, 100)
y_values = f(x_values)

# Plotting the curve
plt.plot(x_values, y_values)

for k in range(1, n +1):
    right_endpoint = a + k * dx        # 0.5, 1.0, 1.5, 2.0
    left_edge = right_endpoint - dx     # left side of that rectangle
    height = f(right_endpoint)         # height comes from f
    plt.bar(left_edge, height, width=dx, align='edge', alpha=0.5, edgecolor='black')

# Display
plt.show()
```
