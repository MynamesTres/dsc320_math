# Working with Matrices 



**Numpy matrix**

1. Enter each of the matrices A, B, and C as Numpy matrix.

```python
# Creating matrices for A,B,C using numpy 

A = np.array([[1, -2, 3, 7],
              [2, 1, 1, 4],
              [-3, 2, -2, 10]])

B = np.array([[0, 1, -3, -2],
              [10, -1, 2, -3],
              [5, 1, -1, 4]])

C = np.array([[4, 0, -2, 3],
              [3, 6, 9, 7],
              [2, 2, 5, 1],
              [9, 4, 6, -2]])

print(f"A\n{A}\n\nB\n{B}\n\n C\n{C}")
```

**Matrices Dimensions**

2. What are the dimensions of each of the matrices A, B, and C? Display these using Python code.

```python
# Display of matrices dimensions 
print(f"A. dimiensions are: {A.shape}")
print(f"B. dimiensions are: {B.shape}")
print(f"C. dimiensions are: {C.shape}")
```

**Matrix operations defined**

3. Determine if each of the following matrix operations is defined. If it is defined, calculate the result
using Python. If it is not defined, explain why it is not defined.   
 
(a) A + B   
(b) AB    
(c) AC    
(d) CT    


```python
# (a) A + B -> defined, since its the same dimensions
calc_ab = A + B 

# Printing calc_ab 
print(f"(a)\n{calc_ab}\n\n")

# (b) AB - is not defined: A is 3x4, while B is 3x4 
# Due to A's columns being (4), which is not equal to B's rows (3), it'll populate as not defined
print("(b) AB is not defined because A has 4 columns and B has 3 rows - they must match\n\n")

# (c) AC is defined: A is 3x4, and C is 4x4 - the inner dimenisons match (4=4)
calc_ac = A @ C 
print(f"(c)\n{calc_ac}\n\n")

# (d) C^T -> Is defined - transposing will always be defined for any matrix
tranpose_c = C.T
print(f"(d)\n{tranpose_c}")
```

**Property of matrix transpositions**

4. Illustrate the following property of matrix transpositions using these matrices: (A + B)^T = A^T + B^T

```python
# Setting up functions 
left_side_ab = (A + B).T
right_side_ab = A.T + B.T 

# Printing left and right side results 
print("Left side (A+B).T:")
print(left_side_ab)

print("\nRight side A.T + B.T:")
print(right_side_ab)

# Printing result 
print("\nAre they equal?", np.array_equal(left_side_ab, right_side_ab))
```

**Property of matrix transpositions**

5. Illustrate the following property of matrix transposition using these matrices (AC)^T = C^TA^T

```python
# Setting up functions 
left_side_ac = (A @ C).T
right_side_ac = C.T @ A.T 

# Printing left and right side results 
print("left side (A @ C).T:") 
print(left_side_ac)

print("\n Right side C.T @ A.T:")
print(right_side_ac)

# Printing result 
print("\nAre they equal?", np.array_equal(left_side_ac, right_side_ac))
```

**Finding C^-1** 

6. Fine C−1 using Python code.

```python
# Setting up the inverse functions 
C_inverse = np.linalg.inv(C)

# Printing result 
print(f" C inverse: \n{C_inverse}")
```

# Working with Images 

For this exercise, you will be working with the image file week7 image.jpg.

**Display the image** 

1. Use the following code to display the image.



```python
# Importing OpenCV library to read images & Matplotlib for image display
import cv2 
import matplotlib.pyplot as plt 

# Importing the image default and OpenCV import image in BGR format 
image_bgr = cv2.imread('week7_image.jpg', cv2.IMREAD_COLOR)

# Converting the image to RGB format 
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# Displaying the image 
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

**Image Resolution** 

2. Find the resolution of this image. Hint: OpenCV imports the image as a Numpy array, where each
pixel corresponds to an entry in the array.


```python
# Printing dimensions of the image 
print(f"{image_rgb.shape}\n")

# Function to grab the dimensions 
height, width, color = image_rgb.shape

# Printing Results 
print(f"Height: {height}, Width: {width}, Color: {color}")
```

**Array dimensions**
     
3. In checking the array dimensions, you should see that three numbers are displayed. What is this third
number, and why is it there?

**Answer** 

- The third dimension represents the color channels. The pixels in the color image aren't stored as a single value, but as three values, each representing Red, Green, and Blue. So this image, instead of being one grid of pixels, is more like three stacked grids, each 453 x 600.
