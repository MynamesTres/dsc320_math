```python
# Importing Libaries 
import pandas as pd 
import matplotlib.pyplot as plt 

# Importing CSV
qb_stats = pd.read_csv('qb_stats.csv')
```

# Summarizing Data

In this exercise, you will be working with the file qb stats.csv. This file contains stats on NFL quar-
terbacks in the 2019 season. The data was obtained from
https://www.footballdb.com/statistics/nfl/player-stats/passing/2019/regular-season. The data includes
the following.

(a) Find the mean of each numerical column of data.

```python
# Finding the mean in the QB stats
mean_data = qb_stats.mean(numeric_only=True)
print(f'Mean Data \n{mean_data}\n')
```

(b) Find the standard deviation of each numerical column of data.

```python
# Finding the standard deviation in the QB stats
std_data = qb_stats.std(numeric_only=True)
print(f'Standard deviation Data\n{std_data}\n')
```

(c) Create a histogram of the number of yards; label it appropriately.

```python
# Creating a histogram of passing years and number of quarterbacks
plt.figure() 
plt.hist(qb_stats['yds'], bins=10, edgecolor='white', color='#0072B2')

# Labels
plt.title('Distribution of QB Passing Yards, 2019 Season')
plt.xlabel('Passing Yards')
plt.ylabel('Number of Quarterbacks')
plt.show()
```

(d) Create a boxplot of the number of touchdowns. Identify any outliers.

```python
# Creating a boxplot of touchdowns
plt.figure() 
plt.boxplot(qb_stats['td'])

# Labels
plt.title('Distribution of QB Touchdowns, 2019 Season')
plt.ylabel('Touchdowns')
plt.show()

# Outliers using the IQR rules
q1 = qb_stats['td'].quantile(0.25)
q3 = qb_stats['td'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = qb_stats[(qb_stats['td'] < lower_bound) | (qb_stats['td'] > upper_bound)]
print(outliers[['name', 'td']])
```

```python
# Importing Libaries 
import pandas as pd 
import matplotlib.pyplot as plt 

# Importing CSV
survey_data = pd.read_csv('survey_data.csv')
count = len(survey_data)
```

# Calculating probabilities from Data 

In this exercise, you will be working with the file survey data.csv. This data has a list of college
students who were asked their hair color and eye color.

(a) Based oﬀ of this data, what is the probability a college student has brown hair?

```python
# Calculation of students who have brown hair 
brown_hair = (survey_data['hair_color'] == 'brown').sum() / count  

print(f"Brown Hair: {brown_hair:.3f}")
```

(b) Based oﬀ of this data, what is the probability a college student has blue eyes?

```python
# Calculation of students who have blue eyes
blue_eyes = (survey_data['eye_color'] == 'blue').sum() / count  

print(f"Blue eyes: {blue_eyes:.3f}")
```

(c) Based oﬀ of this data, what is the probability a college student has blue eyes given that they have
brown hair?

```python
# Calculation of students who have blue eyes that also have brown hair (blue eyes | brown hair)
brown_hair = survey_data[survey_data['hair_color'] == 'brown']
blue_eyes_brown_hair = (brown_hair['eye_color'] == 'blue').sum() / len(brown_hair)

print(f"blue eyes | brown hair: {blue_eyes_brown_hair:.3f}")
```

(d) Based oﬀ of this data, what is the probability a college student has brown hair given that they
have blue eyes?

```python
# Calculation of students who have brown hair that also have blue eyes (brown hair| blue eyes)
blue_eyes = survey_data[survey_data['eye_color'] == 'blue']
brown_hair_blue_eyes = (blue_eyes['hair_color'] == 'brown').sum() / len(blue_eyes)

print(f"brown hair & blue eyes: {brown_hair_blue_eyes}")
```

(e) Do your results above indicate that college students having brown hair and blue eyes are inde-
pendent of one another? Explain.


**Answer**
Blue eyes = 24.7% while [blue eyes | brown hair] = 21.7%, which is a 3-point gap. Since the conditional probability differs from the overall probability, hair color and eye color are not independent. Knowing that someone has brown hair slightly lowers the odds that they will have blue eyes. 

(f) Create a bar graph of the hair color and eye color of this group of students. Label the graphs
appropriately.

```python
# Creating a bar graph for hair color 
survey_data['hair_color'].value_counts().plot(kind='bar', color='#0072B2')

# Label
plt.title('Hair Color Distribution')
plt.xlabel('Hair color')
plt.ylabel('Count')
plt.show() 

# Creating a bar graph for eye color
survey_data['eye_color'].value_counts().plot(kind='bar', color='#0072B2')

# Label
plt.title('Eye Color Distribution')
plt.xlabel('Eye Color')
plt.ylabel('Count')
plt.show()

```
